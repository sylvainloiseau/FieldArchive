```python
import pandas as pd
from fuzzywuzzy import fuzz

fieldArchivePrefix = 'https://fr.cnrs.lacito.fr/fieldArchive/'
matchedWith_col = fieldArchivePrefix + 'matchedWith'
match_status_col = fieldArchivePrefix + 'matchStatus'
rdf_type_col = "rdf:type"

# pivoted (wide) tables, one row per entity
source = pd.read_csv('/content/sample_data/external_datasource.csv')
target = pd.read_csv('/content/sample_data/internal_datasource.csv')

# print(source)

# value_wide = source.pivot(index="entity", columns="property", values="value")
# type_wide  = source.pivot(index="entity", columns="property", values="typevalue")
# type_wide.columns = [f"{col}_type" for col in type_wide.columns]

STRUCTURAL_PROPS = {matchedWith_col, match_status_col}

# ---------------------------------------------------------------------------
# Multivalue-aware pivot
# ---------------------------------------------------------------------------

def pivot_multivalue(df):
    df = df.copy()
    df['entity'] = df['entity'].astype(str).str.strip()
    df['property'] = df['property'].astype(str).str.strip()

    grouped = df.groupby(['entity', 'property'])

    # dedupe repeated identical values (e.g. the same row appearing twice
    # by accident) while preserving order
    def dedupe(values):
        return list(dict.fromkeys(values))

    value_lists = grouped['value'].apply(lambda s: dedupe(list(s))).reset_index()
    type_first = grouped['typevalue'].first().reset_index()

    value_rows = {}
    for _, r in value_lists.iterrows():
        entity, prop, values = r['entity'], r['property'], r['value']
        cell = values[0] if prop in STRUCTURAL_PROPS else values
        value_rows.setdefault(entity, {})[prop] = cell

    type_rows = {}
    for _, r in type_first.iterrows():
        entity, prop, t = r['entity'], r['property'], r['typevalue']
        type_rows.setdefault(entity, {})[prop + '_type'] = t

    entities = sorted(set(value_rows) | set(type_rows))
    records = []
    for e in entities:
        rec = {'entity': e}
        rec.update(value_rows.get(e, {}))
        rec.update(type_rows.get(e, {}))
        records.append(rec)

    return pd.DataFrame(records)

# pivot

df_wide_source = pivot_multivalue(source)
df_wide_target = pivot_multivalue(target)

print(df_wide_source.columns.tolist())
print(df_wide_target.columns.tolist())


def clean_entities_with_same_iri(df_source, df_target):
    # Entities that exist in both dataframes
    common_entities = set(df_source["entity"]).intersection(set(df_target["entity"]))

    print(f"Entities to remove: {common_entities}")

    # Remove common entities from both
    df_source = df_source[~df_source["entity"].isin(common_entities)]
    df_target = df_target[~df_target["entity"].isin(common_entities)]


    return df_source, df_target

def ensure_col(df, col):
    if col not in df.columns:
        df[col] = pd.NA
    return df

def clean_matched_entities(df_source, df_target):
    df_source = ensure_col(df_source.copy(), matchedWith_col)
    df_source = ensure_col(df_source, match_status_col)
    df_target = ensure_col(df_target.copy(), matchedWith_col)

    resolved_mask = df_source[match_status_col].isin(['confirmed', 'manual'])
    matched_target_iris = df_source.loc[resolved_mask, matchedWith_col].dropna().unique()


    df_target = df_target[~df_target['entity'].isin(matched_target_iris)]
    df_source = df_source[~resolved_mask]

    # Entities explicitly excluded from matching: drop from source pool
    excluded_mask = df_source[match_status_col] == 'rejected'

    rejected_pairs = list(
        df_source.loc[excluded_mask, ['entity', matchedWith_col]]
        .dropna(subset=[matchedWith_col])
        .itertuples(index=False, name=None)
    )

    return df_source, df_target, rejected_pairs

df_wide_source, df_wide_target = clean_entities_with_same_iri(df_wide_source, df_wide_target)
df_wide_source, df_wide_target, rejected_pairs = clean_matched_entities(df_wide_source, df_wide_target)

# def compute_similarity(df_source, df_target):
#   list_columns_source = df_source.columns.tolist()
#   list_columns_target = df_target.columns.tolist()
#   for row in df_source.to_dict(orient="records"):

#       print("Entity:", row["entity"])
#       for column in list_columns_source:
#         print(column, " ", row.get(column))
#       print("---")


# print(df_wide_int.columns.tolist())
# print(df_wide_ext.columns.tolist()
for row in df_wide_target.to_dict(orient="records"):
    print("Entity:", row["entity"])

# for row in df_wide_ext.to_dict(orient="records"):
#     print("Entity:", row["entity"])
#     print("Name:", row.get("foaf:givenName"))
#     print("Family:", row.get("foaf:familyName"))
#     print("---")

# candidates_df = pd.DataFrame(candidates).sort_values('name_score', ascending=False)
```

```python
from numpy import nan
from datetime import date
import calendar
import re


def is_missing(val):
    if val is None:
        return True
    if isinstance(val, list):
        return len(val) == 0
    try:
        return bool(pd.isna(val))
    except (TypeError, ValueError):
        return False

def best_score_over_values(source_val, target_val, score_pair_fn):
    """
    source_val / target_val: each a scalar or a list of values.
    score_pair_fn(s, t) -> float score in [0, 100], or None if the pair
    can't be compared at all.
    Returns the best (max) score across every (s, t) combination, or
    None if nothing was comparable.
    """
    if is_missing(source_val) or is_missing(target_val):
        return None

    s_list = source_val if isinstance(source_val, list) else [source_val]
    t_list = target_val if isinstance(target_val, list) else [target_val]

    best = None
    for s in s_list:
        for t in t_list:
            if is_missing(s) or is_missing(t):
                continue
            score = score_pair_fn(s, t)
            if score is not None and (best is None or score > best):
                best = score
    return best


# ---------------------------------------------------------------------------
# 6. Comparators (multivalue-aware)
# ---------------------------------------------------------------------------
def compare_strings(row_source, row_target, column_name, weights):
    val_source, val_target = row_source.get(column_name), row_target.get(column_name)
    best = best_score_over_values(
        val_source, val_target,
        lambda s, t: fuzz.ratio(str(s).lower(), str(t).lower())
    )
    if best is None:
        return 0, 0
    return best * weights[column_name], weights[column_name]


def compare_numbers(val_source, val_target, weight, tolerance=0):
    def score_pair(s, t):
        try:
            n_s, n_t = float(s), float(t)
        except (TypeError, ValueError):
            return None
        return 100 if abs(n_s - n_t) <= tolerance else 0

    best = best_score_over_values(val_source, val_target, score_pair)
    if best is None:
        return 0, 0
    return best * weight, weight


def parse_partial_date(value):
    value = str(value).strip()
    if re.fullmatch(r'\d{4}', value):
        y = int(value)
        return date(y, 1, 1), date(y, 12, 31)
    if re.fullmatch(r'\d{4}-\d{2}', value):
        y, m = map(int, value.split('-'))
        last_day = calendar.monthrange(y, m)[1]
        return date(y, m, 1), date(y, m, last_day)
    if re.fullmatch(r'\d{4}-\d{2}-\d{2}', value):
        y, m, d = map(int, value.split('-'))
        exact = date(y, m, d)
        return exact, exact
    return None, None


def compare_dates(val_source, val_target, weight):
    def score_pair(s, t):
        start_s, end_s = parse_partial_date(s)
        start_t, end_t = parse_partial_date(t)
        if start_s is None or start_t is None:
            return None
        overlaps = max(start_s, start_t) <= min(end_s, end_t)
        if not overlaps:
            return 0
        if start_s == end_s == start_t == end_t:
            return 100
        return 85

    best = best_score_over_values(val_source, val_target, score_pair)
    if best is None:
        return 0, 0
    return best * weight, weight
```

```python

def determine_type_group(source_types, target_types):
    if not source_types or not target_types:
        return True

    source_type_group = []
    target_type_group = []

    for entity_group in ENTITY_EQUIVALENCES["equivalentGroups"]:
        all_equivalent_types = entity_group["equivalentTypes"]
        if not source_type_group and set(source_types).issubset(all_equivalent_types):
            source_type_group = all_equivalent_types
        if not target_type_group and set(target_types).issubset(all_equivalent_types):
            target_type_group = all_equivalent_types
        if source_type_group and target_type_group:
            break

    return entities_have_equivalent_types(source_type_group, target_type_group)


def entities_have_equivalent_types(source_group_types, target_group_types):
    if not source_group_types or not target_group_types:
        return False
    if set(source_group_types) == set(target_group_types):
        return True
    intersection = set(source_group_types) & set(target_group_types)
    if len(intersection) == 0:
        return False
    else :
      return True


def matching_algorithm(df_source, df_target, columns, weights=None, threshold=75):
    candidates = []
    if weights is None:
        weights = {col: 1 / len(columns) for col in columns}

    for row_source in df_source.to_dict(orient="records"):
        for row_target in df_target.to_dict(orient="records"):
            iri_source = row_source["entity"]
            iri_target = row_target["entity"]

            if (iri_source, iri_target) in rejected_pairs:
                continue


            s_types_list = row_source.get("rdf:type") if isinstance(row_source.get("rdf:type"), list) else [row_source.get("rdf:type")]
            t_types_list = row_target.get("rdf:type") if isinstance(row_target.get("rdf:type"), list) else [row_target.get("rdf:type")]

            res = determine_type_group(s_types_list, t_types_list)

            if (not res):
                continue
            print("Type is equivalent ? between : ", iri_source, iri_target, res)

            # print(iri_source, iri_target

            total_score, compared_weight = 0, 0
            for col in columns:
                col_type = row_source.get(col + '_type')

                if col_type == 'string':
                    s, w = compare_strings(row_source, row_target, col, weights)
                    total_score += s
                    compared_weight += w

                elif col_type == 'date':
                    if not is_missing(row_target.get(col)):
                        s, w = compare_dates(row_source[col], row_target[col], weights[col])
                        total_score += s
                        compared_weight += w

                elif col_type == 'number':
                    if not is_missing(row_target.get(col)):
                        s, w = compare_numbers(row_source[col], row_target[col], weights[col])
                        total_score += s
                        compared_weight += w

            if compared_weight == 0:
                continue

            aggregate_score = total_score / compared_weight
            if aggregate_score >= threshold:
                candidates.append({
                    "source_entity": iri_source,
                    "target_entity": iri_target,
                    "rdf_type": row_source["rdf:type"],
                    "score": aggregate_score,
                })

    return candidates


columns = [
    col for col in df_wide_source.columns
    if not col.endswith("_type")
    and col not in ["entity", "rdf:type"]
]

results = matching_algorithm(df_wide_source, df_wide_target, columns)
for r in results:
    print(r)
```
